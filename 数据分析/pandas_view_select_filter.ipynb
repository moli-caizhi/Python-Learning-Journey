```python
# 数据选择
df2 = pd.read_csv("data/sales.csv", usecols = ['订单号', '产品类别', '产品名称', '销售数量', '单价'])

# 1. 选择列
# 1.1 选择单列
df2['产品类别']
# type(df2['产品类别']) # pandas.core.series.Series

# 1.2 选择多列 --> [[]]
df2[[ '产品名称', '单价']]
df2[['产品类别', '产品名称', '单价']]
# type(df2[[ '产品名称', '单价']]) # pandas.core.frame.DataFrame

# index_col: 指定列作为索引
df3 = pd.read_csv("data/sales.csv", usecols = ['订单号', '产品类别', '产品名称', '销售数量', '单价'], index_col='订单号')
df3
# 2. 选择行 --> iloc, loc
# 2.1 iloc --> 基于 行号(不是索引, 是位置, 从0开始) 选择行, 语法: df.iloc[start, stop, step],
# start: 起始下标(包含)
# stop: 结束下标(不包含)
# step: 步长,隔几个取一条
df3.iloc[0:5:1]
df3.iloc[0:5] # 结果同上

# 2.2 loc --> 基于 索引(不是行号, 是索引) 选择行, 语法: df.loc[start, stop, step]
# start: 起始索引(包含)
# stop: 结束索引(包含)
# step: 步长,隔几个取一条
# df3.loc[0:5:1] # 报错, 因为订单号变为索引了, 不是从 0 开始, 需要使用订单号作为索引
df2.loc[0:5:1]
```

```python
df4 = pd.read_csv("data/sales.csv", usecols = ['订单号', '产品类别', '产品名称', '销售数量', '单价'])

# 数据过滤 ---> df[filter]
# 1. 获取 销售数量 >= 10 的订单数据
df4[df4['销售数量'] >= 10]

# 2. 获取产品类别为 食品 或 图书 的订单数据
df4[df4['产品类别'].isin(['食品', '图书'])]

# 3. 获取 单价在100-200之间 的订单数据
df4[(df4['单价'] >= 100) & (df4['单价'] <= 200)]
df4[df4['单价'].between(79, 199)] # --> 默认包含边界
df4[df4['单价'].between(79, 199, inclusive='left')] # --> 不包含右边边界
df4[df4['单价'].between(79, 199, inclusive='right')] # --> 不包含左边边界

# 4. 获取 销售数量 >= 8，并且 单价 >= 100 的订单数据 --> 多条件(并且: &, 或者: |)
df4[(df4['销售数量'] >= 8) & (df4['单价'] >= 100)]

# 5. 获取 产品类别为 服装/食品，支付方式为 支付宝/微信支付 的订单数据
df4[(df4['产品类别'].isin(['服装', '食品'])) & (df4['支付方式'].isin(['支付宝', '微信支付']))]

```
