```python
#----------------------------------------子图--------------------------------------------
import matplotlib.pyplot as plt
from matplotlib.axes import Axes
# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

# 创建子图
# nrows: 行数, ncols: 列数, figsize: 图像大小(宽, 高), dpi: 图像分辨率
# figure: 画布对象, axes: 子图数组
figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)

# 图1: 柱状图(世界石油储备) --> 适合展示 数量对比 --> bar
countries = ["美国", "中国", "英国", "日 本", "德国"]
values = [190, 160, 150, 140, 130]

# axes1 是 Axes 类型, 没有给他类型, PyCharm 不会提示方法, 进行类型注解会让代码有丰富的提示
axes1:Axes = axes[0] # 第一个子图
# axes1 # --> <Axes: >
axes1.bar(countries, values, width=0.6, color="green") # 柱状图, width: 柱子宽度, color: 柱子颜色
axes1.set_title("世界石油储备", fontsize=16) # 设置标题
axes1.set_xlabel("国家", ) # 设置X轴标签
axes1.set_ylabel("石油储备(亿吨)",) # 设置Y轴标签
axes1.grid(linestyle="--", alpha=0.3) # 调用 grid 前面不需要 set

# 图2: 饼状图(世界人口) --> 适合展示比例构成 --> pie
cuntries2 = ['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
values = [14.51, 14.09, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]
axes2:Axes = axes[1] # 第二个子图
# 第一个 % 是格式占位符开头，.1f 表示保留 1 位小数，末尾两个 %%` 转义后只输出一个百分号
axes2.pie(values, labels=cuntries2, autopct='%1.1f%%') # 先传递值, 后传递标签
axes2.set_title("世界人口比例", fontsize=16) # 设置标题
axes2.legend(loc='lower center', ncol=5, bbox_to_anchor=(0.5, -0.1)) # ncol=5, 设置图例列数为5, bbox_to_anchor(控制水平位置, 控制垂直位置):  设置图例位置

# 保存图片
plt.savefig("data/chart.png")

plt.show() # 显示图形
```
