```python
#------------------------------------------数据清洗-----------------------------------------
import pandas as pd

# 1. 读取数据
df = pd.read_csv("data/sales.csv")
df

# 2. 数据清洗
# 2.1 处理缺失值
df.isnull() # 检查缺失值，返回布尔值矩阵

# 2.1.1 删除缺失值
df.dropna() # 删除包含缺失值的行 --> 有返回值, 为 DataFrame
df = df.dropna(axis=1) # 删除包含缺失值的列, axis 默认为 0
df
# 2.1.2 填充缺失值 --> 有返回值, 为 DataFrame
df.fillna("--") # 填充缺失值
df.ffill() # 填充缺失值, 用前一个值填充
df.bfill() # 填充缺失值, 用后一个值填充

```

```python
# 2.2 重复值的处理
# 2.2.1 查看重复值
df.duplicated() # 检查重复值(所有的列的数据都重复)，返回布尔值向量
df.duplicated(subset=["订单号"]) # 检查重复值(只检查订单号这一列的数据是否重复)，返回布尔值向量

# 2.2.2 删除重复值
df.drop_duplicates(subset=["订单号"]) # keep="first" 根据订单号这一列删除重复行, 保留第一次出现的那一行, 删掉后续重复的行 --> keep 默认为 first
df.drop_duplicates(subset=["订单号"], keep="last") # 保留最后一次出现的那一行, 删掉前面出现的行
df.drop_duplicates(subset=["订单号"], keep=False) # 删除所有重复行
```

```python
# 2.3 异常值处理
# 2.3.1 查看异常值
df[df["单价"] < 0]

# 2.3.2 删除异常值
df.drop(df[df["单价"] < 0].index)

# 2.3.3 修复异常值
df["单价"] = df["单价"].abs() # 绝对值
df
```

```python
# 2.4 数据格式处理
df["订单日期"] = df["订单日期"].replace("2025/06/02", "2025-06-02") # 整体替换, 不推荐, 不能处理大量数据
df

df["订单日期"] = df["订单日期"].str.replace("/", "-") # .str 是 pandas 的字符串访问器, 用于字符串操作
df
```
