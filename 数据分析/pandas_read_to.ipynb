```python
#---------------------------------------数据的读取与写入-------------------------------------
import pandas as pd

# 读取数据 --> read_csv
# usecols: 选择需要的列, cols 是 columns 的缩写
df = pd.read_csv("data/sales.csv", usecols = ['订单号', '产品类别', '产品名称', '销售数量', '单价'])

# 数据处理
# df[]: 若列表中没有指定的列名, 则会报错
# df[]: 若列表中没有指定的列名, 进行列表中已有列的运算则会添加新的列
df["销售金额"] = df["销售数量"] * df["单价"]

# 写入数据 --> to_csv, index=False: 不写入行索引
df.to_csv("data/sales_01.csv", index=False)
```
